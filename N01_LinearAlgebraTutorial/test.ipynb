```python
import numpy as np 

```

```python
A = np.array([[1, 2], [3, 4], [5, 6]], dtype=float)

# 1. 奇异值分解 SVD
U, s, Vt = np.linalg.svd(A)
print("左奇异矩阵 U:\n", U)
print("奇异值 (Singular values):", s)
print("右奇异矩阵 Vᵀ:\n", Vt)

# 2. 计算 Moore-Penrose 广义伪逆 A⁺
A_pinv = np.linalg.pinv(A)
print("广义伪逆 A⁺:\n", A_pinv)
```
