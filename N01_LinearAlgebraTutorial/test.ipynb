```python
A = np.array([[1, 2], [3, 4], [5, 6]], dtype=float)

# 1. 奇异值分解 SVD
U, s, Vt = np.linalg.svd(A)
print("左奇异矩阵 U:\n", U)
print("奇异值 (Singular values):", s)
print("右奇异矩阵 Vᵀ:\n", Vt)

# 2. 计算 Moore-Penrose 广义伪逆 A⁺
A_pinv = np.linalg.pinv(A)
print("广义伪逆 A⁺:\n", A_pinv)
```

```python
np.random.seed(42)
mean = [0, 0, 0]
cov_true = np.array([[4.0, 1.8, 0.2],
                     [1.8, 2.0, 0.5],
                     [0.2, 0.5, 1.0]])  # 对称正定协方差矩阵
X = np.random.multivariate_normal(mean, cov_true, size=1000)

# 2. 中心化并计算样本协方差矩阵
X_centered = X - np.mean(X, axis=0)
Sigma = (X_centered.T @ X_centered) / (len(X) - 1)

# 3. 对实对称矩阵调用专属高精度算法 eigh (返回升序特征值和正交特征向量)
eigenvalues, Q = np.linalg.eigh(Sigma)

# 倒序排列：使得最大特征值排在最前
idx = np.argsort(eigenvalues)[::-1]
evals_sorted = eigenvalues[idx]
Q_sorted = Q[:, idx]

print("=== 样本协方差矩阵谱分解结果 ===")
print("各主轴特征值 (方差贡献):", np.round(evals_sorted, 4))
explained_variance_ratio = evals_sorted / np.sum(evals_sorted)
print("方差解释率占比:        ", np.round(explained_variance_ratio * 100, 2), "%")

# 4. 谱分解重构矩阵验证: A = Q @ diag(evals) @ Q^T
Sigma_reconstructed = Q_sorted @ np.diag(evals_sorted) @ Q_sorted.T
reconstruction_err = np.linalg.norm(Sigma - Sigma_reconstructed)
print(f"外积谱分解重构误差 ||A - QΛQ^T||: {reconstruction_err:.2e} (完全精确！")
print(Q)
```
